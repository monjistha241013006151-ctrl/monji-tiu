Problem 3: Checking Routes Between Cities Using DFS
Problem Statement:
A small road network connects several cities. Write a Python program using DFS that:
1. Checks whether a route exists between two given cities.
2. Finds all separate groups of cities that are connected to each other but not to any
other group (these are called connected components).

```python
# Road network represented using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A'],
    'D': ['B'],
    'E': ['F'],
    'F': ['E'],
    'G': []
}


# Function to check if a route exists using DFS
def route_exists(graph, start, destination, visited=None):

    if visited is None:
        visited = set()

    # If we reach the destination
    if start == destination:
        return True

    visited.add(start)

    # Visit all neighbouring cities
    for neighbour in graph[start]:
        if neighbour not in visited:
            if route_exists(graph, neighbour, destination, visited):
                return True

    return False


# Function to find one connected component
def find_component(graph, start, visited):
    component = []

    visited.add(start)
    component.append(start)

    for neighbour in graph[start]:
        if neighbour not in visited:
            component.extend(find_component(graph, neighbour, visited))

    return component


# Function to find all connected components
def connected_components(graph):
    visited = set()
    components = []

    for city in graph:
        if city not in visited:
            component = find_component(graph, city, visited)
            components.append(component)

    return components


# Check routes
print("Route from A to D:",
      route_exists(graph, 'A', 'D'))

print("Route from A to F:",
      route_exists(graph, 'A', 'F'))

# Find connected components
components = connected_components(graph)

print("\nConnected Components:")
for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
