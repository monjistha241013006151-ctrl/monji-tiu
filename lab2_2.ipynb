```python
Problem 2: The Water Jug Problem (Solved using BFS)
Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.
```

```python
from collections import deque

# Capacities of the two jugs
CAP_A = 4
CAP_B = 3


def get_next_states(a, b):
    states = []

    # 1. Fill Jug A
    states.append((CAP_A, b))

    # 2. Fill Jug B
    states.append((a, CAP_B))

    # 3. Empty Jug A
    states.append((0, b))

    # 4. Empty Jug B
    states.append((a, 0))

    # 5. Pour A -> B
    amount = min(a, CAP_B - b)
    states.append((a - amount, b + amount))

    # 6. Pour B -> A
    amount = min(b, CAP_A - a)
    states.append((a + amount, b - amount))

    return states


def bfs():
    start = (0, 0)

    # Queue stores (state, path)
    queue = deque([(start, [start])])

    visited = set()
    visited.add(start)

    while queue:
        (a, b), path = queue.popleft()

        # Goal condition
        if a == 2 or b == 2:
            return path

        # Generate next states
        for next_state in get_next_states(a, b):

            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None


# Find minimum solution
solution = bfs()

print("Minimum sequence of states:")

for i, state in enumerate(solution):
    print(f"Step {i}: Jug A = {state[0]} L, Jug B = {state[1]} L")
```
